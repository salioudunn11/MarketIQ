```python
import pandas as pd
import numpy as np
```

```python
# Load Raw GDPC1 File
df_gdp = pd.read_csv('GDPC1.csv')
```

```python
# Clean Header Whitespace & Apply Clear Team Column Names
df_gdp.columns = df_gdp.columns.str.strip()

df_gdp = df_gdp.rename(columns={
    'observation_date': 'Date',
    'OBSERVATION_DATE': 'Date',
    'GDPC1': 'Real_GDP_$BUSD',
    'gdpc1': 'Real_GDP_$BUSD'
})
```

```python
# Sort Chronologically
df_gdp = df_gdp.sort_values('Date').reset_index(drop=True)
```

```python
print(df_gdp.head())
```

```python
# Convert Date to datetime format
df_gdp['Date'] = pd.to_datetime(df_gdp['Date'], errors='coerce')
```

```python
# Drop rows with invalid dates and sort the DataFrame by date
df_gdp = df_gdp.dropna(subset=['Date']).sort_values('Date').reset_index(drop=True)
```

```python
# Handle missing values (Forward fill)
df_gdp['Real_GDP_$BUSD'] = pd.to_numeric(df_gdp['Real_GDP_$BUSD'], errors='coerce').ffill()
```

```python
# Create Year-Month string feature (e.g., '1947-01')
df_gdp['Year_Month'] = df_gdp['Date'].dt.strftime('%Y-%m')
```

```python
df_gdp_clean = df_gdp[['Date', 'Year_Month', 'Real_GDP_$BUSD']]
```
