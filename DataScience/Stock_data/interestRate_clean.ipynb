```python
import pandas as pd
import numpy as np

df = pd.read_csv('../FRED/interest_rate.csv')

df
```

```python
#remove any row that contains null values 
df_clean = df.dropna().reset_index(drop=True)
```

```python
print("Remaining null values:\n", df_clean.isnull().sum())
```

```python

# 1. Replace common string placeholders with true NaN values
df = df.replace(['.', 'N/A', 'nan', 'null', ' '], np.nan)

# 2. Drop all rows containing any NaN values
df_clean = df.dropna().reset_index(drop=True)

# 3. Print verification info
print(f"Original row count: {len(df)}")
print(f"Cleaned row count:  {len(df_clean)}")
print(f"Rows dropped:       {len(df) - len(df_clean)}")
```

```python
# Drops any column that contains missing values
df_clean = df.dropna(axis=1)
```

```python
# Rename columns ('observation_date' or 'DATE' -> 'Date', 'DFF' -> 'Interest_Rate')
rename_dict = {}
for col in df.columns:
    if col.lower() in ['observation_date', 'date']:
        rename_dict[col] = 'Date'
    elif col.upper() in ['DFF', 'VALUE', 'INTEREST_RATE', 'INTEREST RATE']:
        rename_dict[col] = 'Interest_Rate'
```

```python
df = df.rename(columns=rename_dict)
```

```python
# Convert types: strings like '.' become NaN, and dates parse cleanly
df['Date'] = pd.to_datetime(df['Date'])
df['Interest_Rate'] = pd.to_numeric(df['Interest_Rate'], errors='coerce')
```

```python
#  Filter out weekends (0=Monday ... 4=Friday, 5=Saturday, 6=Sunday)
df_weekday = df[df['Date'].dt.dayofweek < 5]
```

```python
# Remove all remaining null values and sort chronologically
df_clean = df_weekday.dropna().sort_values('Date').reset_index(drop=True)
```

```python
# Export the clean file
df_clean.to_csv('dff_clean.csv', index=False)
```

```python
print(f"Original rows: {len(df)}")
print(f"Cleaned weekday rows: {len(df_clean)}")
```
