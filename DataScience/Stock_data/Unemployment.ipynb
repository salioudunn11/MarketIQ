```python
import pandas as pd
import numpy as np
```

```python
df_unrate = pd.read_csv('unrate.csv')
```

```python
df_unrate.columns = df_unrate.columns.str.strip()  # Strip whitespace from column names
df_unrate = df_unrate.dropna()  # Drop rows with missing values
df_unrate = df_unrate.drop_duplicates()  # Drop duplicate rows
df_unrate = df_unrate.rename(columns={
    'observation_date': 'Date',
    'UNRATE': 'Unemployment_Rate'
})
```

```python
#Datetime conversion
df_unrate['Date'] = pd.to_datetime(df_unrate['Date'], errors='coerce')

# chronological sorting
df_unrate = df_unrate.sort_values('Date').reset_index(drop=True)
```

```python
# Handle Missing Values (Forward fill for 2025-10-01 missing reading)
df_unrate['Unemployment_Rate'] = pd.to_numeric(df_unrate['Unemployment_Rate'], errors='coerce').ffill()
```

```python
#  Extract Year-Month Feature ('YYYY-MM')
df_unrate['Year_Month'] = df_unrate['Date'].dt.strftime('%Y-%m')
```
