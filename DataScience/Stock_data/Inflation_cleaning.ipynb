```python
import pandas as pd
import numpy as np

# Load raw CPI dataset
df = pd.read_csv('CPIAUCSL.csv')
```

```python
# Clean column headers
df.columns = df.columns.str.strip()

# Rename columns
df = df.rename(columns={
    'observation_date': 'Date',
    'OBSERVATION_DATE': 'Date',
    'CPIAUCSL': 'CPI',
    'cpiaucsl': 'CPI'
})
```

```python
# Convert Date to datetime format
df['Date'] = pd.to_datetime(df['Date'])
df = df.sort_values('Date').reset_index(drop=True)
```

```python
#Handle missing value at 2025-10-01 using forward fill
df['CPI'] = df['CPI'].ffill()
```

```python
# Feature Engineering: YoY Inflation Rate & MoM Change
# YoY = % change compared to 12 months ago
df['CPI_YoY_Inflation'] = (df['CPI'].pct_change(12) * 100).round(4)
```

```python
# MoM = % change compared to previous month
df['CPI_MoM_Change'] = (df['CPI'].pct_change(1) * 100).round(4)
```

```python
# Round raw CPI to 3 decimal places
df['CPI'] = df['CPI'].round(3)
```
