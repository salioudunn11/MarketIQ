```python
import pandas as pd
import numpy as np

# Load raw CPI dataset
df = pd.read_csv('CPIAUCSL.csv')
```

```python
# Clean column headers
df.columns = df.columns.str.strip()

# Rename columns
df = df.rename(columns={
    'observation_date': 'Date',
    'OBSERVATION_DATE': 'Date',
    'CPIAUCSL': 'Consumer_Price_Index',
    'cpiaucsl': 'Consumer_Price_Index',
    'CPI': 'Consumer_Price_Index'
})
```

```python
# Convert Date to datetime format
df['Date'] = pd.to_datetime(df['Date'], errors='coerce')
df = df.sort_values('Date').reset_index(drop=True)
```

```python
#Handle missing value at 2025-10-01 using forward fill
df['Consumer_Price_Index'] = df['Consumer_Price_Index'].ffill()
```

```python
# 1. Ensure Consumer_Price_Index is numeric first
df['Consumer_Price_Index'] = pd.to_numeric(df['Consumer_Price_Index'], errors='coerce')

# 2. Forward fill missing values (e.g., the missing reading at 2025-10-01)
df['Consumer_Price_Index'] = df['Consumer_Price_Index'].ffill()

# 3. Create a Year-Month column (formatted as 'YYYY-MM', e.g., '2025-10')
df['Year_Month'] = df['Date'].dt.strftime('%Y-%m')
```
