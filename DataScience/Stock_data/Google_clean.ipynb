```python
import pandas as pd
import numpy as np

# Load raw GOOG dataset
df = pd.read_csv('GOOG.csv')
```

```python
# Standardize column names (remove space in 'Adj Close')
df = df.rename(columns={'Adj Close': 'Adj_Close'})
```

```python
# Sort chronologically
df = df.sort_values('Date').reset_index(drop=True)
```

```python
# Filter date range to align with macro data (e.g., 2015 onwards)
df = df[df['Date'] >= '2015-01-01'].reset_index(drop=True)
```

```python
# 7. Round price columns to 4 decimal places
price_cols = ['Open', 'High', 'Low', 'Close', 'Adj_Close']
df[price_cols] = df[price_cols].round(4)
```

```python
# Daily Percentage Return & 60-Day Moving Average
df['Daily_Return'] = df['Close'].pct_change().round(6)
df['SMA_60'] = df['Close'].rolling(window=60).mean().round(4)
```

```python
# Remove NaNs generated by rolling calculations
df_clean = df.dropna().reset_index(drop=True)
```

```python
# Save clean CSV
df_clean.to_csv('goog_clean.csv', index=False)
```
